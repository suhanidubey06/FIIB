```python
#byjus case
import pandas as pd
import numpy as np

df = pd.read_csv("/Lecture2_data - Lecture2_data.csv.csv")
```

```python
# Decision making structures
#If else
revenue = df["Revenue_Lakh"].mean()

print("Average Revenue:", revenue)

if revenue > 300:
  print("Average revenue is high")
else:
  print("Average revenue is low")
```

```python
#decision making structures
# Nested If Statement in Python
revenue = df["Revenue_Lakh"].mean()
rating = df["Customer_Rating"].mean()

if revenue >= 300:
  print("Revenue is high")

  if rating >= 4:
     print("Customer satisfaction is also good")
  else:
     print("Customer satisfaction needs improvement")
else:
  print("Revenue is low")
```

```python
# Find the average customer rating
rating = df["Customer_Rating"].mean()

# Make a decision based on the rating
if rating >= 4.5:
    print("Excellent")

elif rating >= 4.0:
    print("Good")

elif rating >= 3.0:
    print("Average")

else:
    print("Needs Improvement")
```

```python
# For Loop
for rating in df["Customer_Rating"]:

  if rating >= 4:
     print("Good Rating")
  else:
     print("Needs Improvement")
```

```python
# For Loop with range() iloc= integer location

for i in range(10):
  print(df["Product"].iloc[i], df["Revenue_Lakh"].iloc[i])
```

```python
#WHILE Loop in Python
#While loop and i,j is variable to track
i = 0

while i < 5:
    revenue = df["Revenue_Lakh"].iloc[i]

    if revenue >= 300:
      print("High Revenue:", revenue)
    else:
      print("Low Revenue:", revenue)

    i = i + 1
```

```python
# LIST: the Starbuks menu (8 products)
menu = ["Cappuccino", "Latte",  "Americano", "Cold Brew",
        "Frappuccino", "Masala Chai", "Blueberry Muffin", "Veg Sandwich"]

# TUPLE: one store's fixed details (StoreID, name, city)
store = ("S01", "Connaught Place", "Delhi")

# SET: cities where Starbucks sells (the duplicate 'Delhi' is removed automatically)
cities = {"Delhi", "Mumbai", "Bengaluru", "Hyderabad", "Pune", "Delhi"} # Changed to a set literal for clarity

# DICTIONARY: product -> base price in Rs
price = {"Cappuccino": 250, "Latte": 270, "Americano": 220, "Cold Brew": 300, "Frappuccino": 350, "Masala Chai": 200, "Blueberry Muffin": 180, "Veg Sandwich": 240}

print("List :", menu)
print("Tuple :", store)
print("Set :", cities, " <- 'Delhi' typed twice, stored once")
print("Dictionary:", price)
print()
print(type(menu), type(store), type(cities), type(price))

```
